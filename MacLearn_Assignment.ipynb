- Country: index of Country

- AdultMortality: Adult Mortality Rates of both sexes (probability of dying between 15 and 60 years per 1000 population)
- AdultMortality-Female: Adult Mortality Rates of Female (probability of dying between 15 and 60 years per 1000 population)
- AdultMortality-Male: Adult Mortality Rates of Male (probability of dying between 15 and 60 years per 1000 population)
- SLS: Number of Short Life Span per 1000 population
- Alcohol: Alcohol, recorded per capita (15+) consumption (in litres of pure alcohol)
- PercentageExpenditure: Expenditure on health as a percentage of Gross Domestic Product per capita(%)
- Measles: Measles - number of reported cases per 1000 population
- BMI: Average: Body Mass Index of entire population
- Under5LS: Number of under-five Life Span per 1000 population
- Polio: Polio (Pol3) immunization coverage among 1-year-olds (%)
- TotalExpenditure: General government expenditure on health as a percentage of total government expenditure (%)
- Diphtheria: Diphtheria tetanus toxoid and pertussis (DTP3) immunization coverage among 1-year-olds (%)
- HIV-AIDS: Factor indicating between 0-4 years life span per 1000 HIV/AIDS 
- GDP: Gross Domestic Product per capita (in USD)
- Population: Population of the country
- Thinness1-19years: Prevalence of thinness among children and adolescents for Age 10 to 19 (% )
- Thinness5-9years: Prevalence of thinness among children for Age 5 to 9(%)
- IncomeCompositionOfResources : Human Development Index in terms of income composition of resources (index ranging from 0 to 1)
- Schooling: Schooling indicator

### Data Splitting

The scikit-learn Python machine learning library provides an implementation of the train-test splitting via function `train_test_split()`. Lets use this to randomly split our data to 80% train set and 20% test set.

```python
from sklearn.model_selection import train_test_split

with pd.option_context('mode.chained_assignment', None):
    dataTrainFrame, dataTestFrame, yTrain, yTest = train_test_split(data, y, test_size=0.2, shuffle=True)


print("Number of instances in the original dataset is {}. After spliting Train has {} instances and test has {} instances."
      .format(data.shape[0], dataTrainFrame.shape[0], dataTestFrame.shape[0]))
```

### Feature Scaling 

To do feature scaling let's understand the data, how much is it range and based on it how can we do feature scaling.

```python
dataTrainFrame.describe()
```

```python
for column in dataTrainFrame.columns:
    col_range = dataTrainFrame[column].max() - dataTrainFrame[column].min()
    maximum = dataTrainFrame[column].max()
    minimum = dataTrainFrame[column].min()
    print(f'Range of {column}: {col_range} , Maximum: {maximum} , Minimum: {minimum}')
```

In our preprocessing phase, the few columns are being excluded from both scaling and normalization procedures. This decision was made based on the criteria that unique identifiers, which serve as arbitrary labels rather than quantitative measures, do not benefit from these transformations. Out of these, there are 4 columns that need to be excluded from scaling or Normalization, these are : 
- ID: which is aUnique Identifier
- Country: which is Nominal
- Year: which is Nominal
- Status: which is Binary

```python
from sklearn.preprocessing import StandardScaler

exclude_columns = ['ID', 'Country', 'Year', 'Status']
cols_to_normalize = [col for col in dataTrainFrame.columns if col not in exclude_columns]

scaler = StandardScaler()

# Fit the scaler on the training data
scaler.fit(dataTrainFrame[cols_to_normalize])

# Now you can transform both your training and test data
dataTrainFrame[cols_to_normalize] = scaler.transform(dataTrainFrame[cols_to_normalize])
dataTestFrame[cols_to_normalize] = scaler.transform(dataTestFrame[cols_to_normalize])
```

### Regularisation

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_error

# Assume X_train, X_test, y_train, y_test are already defined and preprocessed

# Create ridge regression with an alpha value (regularization strength)
ridge_reg = Ridge(alpha=1.0)

# Fit the model on the training data
ridge_reg.fit(X_train, y_train)

# Evaluate the model using the test data
y_pred = ridge_reg.predict(X_test)
mse = mean_squared_error(y_test, y_pred)

# Perform cross-validation to see how well it generalizes
scores = cross_val_score(ridge_reg, X_train, y_train, scoring='neg_mean_squared_error', cv=5)

# Print the Mean Squared Error and the cross-validation results
print(f"Mean Squared Error: {mse}")
print(f"Cross-Validation Scores: {-scores.mean()}")  # Scores are negative, take the negative to make them positive
```

```python
from sklearn.linear_model import Lasso

# Create lasso regression with an alpha value
lasso_reg = Lasso(alpha=0.1)

# Fit the model
lasso_reg.fit(X_train, y_train)

# Evaluate the model
y_pred = lasso_reg.predict(X_test)
mse = mean_squared_error(y_test, y_pred)

# Perform cross-validation
scores = cross_val_score(lasso_reg, X_train, y_train, scoring='neg_mean_squared_error', cv=5)

# Print the results
print(f"Mean Squared Error: {mse}")
print(f"Cross-Validation Scores: {-scores.mean()}")
```
